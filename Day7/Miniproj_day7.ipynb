```python
import pandas as pd
import numpy as np

from sentence_transformers import SentenceTransformer
import chromadb

print("All libraries imported successfully!")
print(f"ChromaDB version: {chromadb.__version__}")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded successfully")
print(f"Model produces vectors of size {model.get_sentence_embedding_dimension()} dimentions")
```

```python
emp_df = pd.read_csv("emp_track.csv")
display(emp_df)
print(f"\nShape: {emp_df.shape}")
print(f"\nColumns: {emp_df.columns.tolist()}")
```

```python
first_note = emp_df.iloc[1]

print(f"Note ID: {first_note['emp_id']}")
print(f"Subject: {first_note['job']}")
print(f"Topic: {first_note['position']}")
print(f"Content: {first_note['job_description']}")
```

```python
# KEYWORD SEARCH

query_keyword = "secure works"
print("="*60)
print(f"KEYWORD SEARCH for: {query_keyword}")
print("="*60)

for i,doc in enumerate(emp_df['job_description']):
    if query_keyword.lower() in doc.lower():
      print((f" FOUND [doc_{i}]: {doc[:80]}..."))
    else:
      print(f" MISSED [doc_{i}]: {doc[:80]}...")
```

```python
all_documents = emp_df['job_description'].tolist()
all_ids = emp_df['emp_id'].tolist()
all_metadatas = [
    {"subject": row['job'], "topic": row['position']}
    for _, row in emp_df.iterrows()
]

embeddings = model.encode(all_documents).tolist()
print(f"Number of documents: {len(all_documents)}")
print(f"Number of IDs: {len(all_ids)}")
print(f"Number of metadatas: {len(all_metadatas)}")
print(f"\nSample ID: {all_ids[0]}")
print(f"Sample metadata: {all_metadatas[0]}")
print(f"Sample document: {all_documents[0][:80]}...")
```

```python
# SEMANTIC SEARCH
from sentence_transformers import util

print(f"Query: {query_keyword}\n")
for i,doc in enumerate(all_documents):
  print(f"Doc{i}: {doc[:80]}...")
  embedding_query = model.encode(query_keyword.lower())
  embedding_doc = model.encode(doc.lower())
  similarity = util.cos_sim(embedding_query, embedding_doc).item()
  if similarity > 0.2:
    print(f" FOUND [doc_{i}]: {doc[:80]}... (similarity={similarity:.2f})\n")
  else:
    print(f" MISSED [doc_{i}]: {doc[:80]}... (similarity={similarity:.2f})\n")
```

```python
client =chromadb.EphemeralClient()
emp_collection=client.get_or_create_collection("emp_track")
print("ChromaDB client created (in-memory mode)")


print(f"Collection name:emp_track ")
print(f"Documents in collection :{emp_collection.count()}")
```

```python
embeddings = model.encode(all_documents).tolist()

emp_collection.add(
    documents=all_documents,
    embeddings=embeddings,
    ids=all_ids,
    metadatas=all_metadatas
)

print("Documents added successfully")
print(f"Number of documents in collection: {emp_collection.count()}")
```

```python
query="Who's work is to secure data?"

results=emp_collection.query(
    query_texts=[query],
    n_results=5
)

print("RESULT KEYS AVAILABLE:")
print(list(results.keys()))
```

```python
print(f"Query")
print("="*60,'\n')

matched_docs = results['documents'][0]
matched_ids = results['ids'][0]
matched_distances = results['distances'][0]
matched_metadata = results['metadatas'][0]

for rank, (doc, doc_id, dist, meta) in enumerate(zip(matched_docs, matched_ids, matched_distances, matched_metadata)):
  print(f"Rank: {rank} | ID: {doc_id} | Distance: {dist:.2f}")
  print(f"Subject: {meta['subject']} | Topic: {meta['topic']}")
  print(f"Document:{doc}\n")
```

```python
filtered_results = emp_collection.query(
    query_texts=[query],
    n_results=3,
    where={"topic": "DevOps Engineer"}
)

print(f"FILTERED QUERY: {query}")
print("Filter only Machine Learning documents")
print("="*60)

for rank, (doc, dist, meta) in enumerate(zip(
    filtered_results['documents'][0],
    filtered_results['distances'][0],
    filtered_results['metadatas'][0]),start=1):
  print(f"Rank: {rank} | Distance: {dist:.2f} | Subject: {meta['subject']}")
  print(f"{doc}\n")
```

```python
print("DISTANCE TO SIMILARITY CONVERSION")
print(f"{'Distance':<15}{'Similarity':<15}{'Interpretation':<20}")
distances=[0.05,0.20,0.40,0.65,0.90]
interpretations=["Near identical","Very similar","Related","Somewhat related","Not related"]
for dist,interp in zip(distances,interpretations):
  similarity=1-dist
  print(f"{dist:<15.2f}{similarity:<15.2f}{interp:<20}")
```
