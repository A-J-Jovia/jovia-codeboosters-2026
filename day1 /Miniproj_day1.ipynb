```python
import pandas as pd

print("="*60)
print(" STUDENT PERFORMANCE EXPLORER REPORT")
print("="*60,"\n")

df = pd.read_csv('student_performance.csv');

print("[1] DATASET OVERVIEW\n-----------------------------------------------------")
print("Total Students: ",df.shape[0])
print("Total Columns: ",df.shape[1])
print("Department: ",len(df['department'].unique()),"\n")  # or use nunique() (no need len())

print("[2] DEPARTMENT-WISE STUDENT COUNT\n-----------------------------------------------------")
print(df['department'].value_counts(),"\n") # or use print(df.groupby('department')['department'].count())

print("[3] SCORE ANALYSIS\n-----------------------------------------------------")
print("Highest Math Score: ",df['math_score'].max())
print("Lowest Math Score: ",df['math_score'].min())
print("Average Math Score: ",df['math_score'].mean().round(2))
print("Average Programming Score: ",df['programming_score'].mean())
print("Average Attendance: ",df['attendance_percentage'].mean(),"\n")

print("[4] GENDER-WISE AVERAGE SCORE\n-----------------------------------------------------")
df['total_score']=(df['math_score']+df['programming_score']+df['science_score']+df['english_score'])
print(df.groupby('gender')['total_score'].mean().round(2),"\n")

print("[5] TOP PERFORMERS\n-----------------------------------------------------")
df['avg_score'] = df['total_score'] / 4
for index, row in df.iterrows():
    if row['avg_score'] > 90:
        print("Name:", row['name'], "| Avg Score:", row['avg_score'])
"""
print("\n[5] TOP PERFORMERS\n-----------------------------------------------------")
df['avg_score']=df['total_score']/4
print(df[df['avg_score'] >90],"\n")
"""

print("\n[6] STUDENTS WHO NEED ATTENDANCE\n-----------------------------------------------------")
low_attendance = df[df['attendance_percentage'] < 75][['name', 'department', 'attendance_percentage']]
if len(low_attendance)>0:
  print(low_attendance.to_string(index=False))
else:
  print("All students have attendance above 75%")
print("\n" + "="*60)
print("Report Complete")
print("="*60)
```
